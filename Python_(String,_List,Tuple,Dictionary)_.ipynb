```python
# 15. Python program to reverse a given string (5 different ways).
s = "Python"

# 1. Slicing
print(s[::-1])

# 2. reversed()
print("".join(reversed(s)))

# 3. Loop
rev = ""
for ch in s:
    rev = ch + rev
print(rev)

# 4. While loop
i = len(s) - 1
while i >= 0:
    print(s[i], end="")
    i -= 1
print()

# 5. Recursion
def reverse(str):
    if len(str) == 0:
        return str
    return reverse(str[1:]) + str[0]

print(reverse(s))
```

```python
# 16. Python program to reverse a string using stack and reversed method.
s = "Python"

# Using stack
stack = list(s)
rev = ""
while stack:
    rev += stack.pop()
print(rev)

# Using reversed()
print("".join(reversed(s)))
```
